## 1. `list`: список товаров

`list` — упорядоченная изменяемая коллекция.

Пример каталога:


````python
products = ["Ноутбук", "Мышь", "Клавиатура", "Монитор", "Наушники"]

print(products)
print(type(products))
````

### Индексы

Индекс — позиция элемента в списке.

Важно:
- первый элемент имеет индекс `0`;
- последний можно получить индексом `-1`.


````python
products = ["Ноутбук", "Мышь", "Клавиатура", "Монитор", "Наушники"]

print("Первый товар:", products[0])
print("Второй товар:", products[1])
print("Последний товар:", products[-1])
````

### Срезы списка

Срез позволяет получить **часть списка**.

Шаблон:

```python
list[start:stop:step]
```

Граница `stop` **не включается**.

Примеры:


````python
products = ["Ноутбук", "Мышь", "Клавиатура", "Монитор", "Наушники"]

print("Первые три:", products[:3])
print("Со второго по четвёртый:", products[1:4])
print("С третьего до конца:", products[2:])
print("Каждый второй:", products[::2])
````

### Список в обратном порядке

Самый короткий способ — срез с шагом `-1`:

```python
products[::-1]
```

Исходный список при этом не изменяется.


````python
products = ["Ноутбук", "Мышь", "Клавиатура", "Монитор", "Наушники"]

reversed_products = products[::-1]

print("Исходный список:", products)
print("Обратный порядок:", reversed_products)
````

### Основные операции со списком

На этом уроке используем:

- `append(value)` — добавить в конец;
- `insert(index, value)` — вставить по индексу;
- `remove(value)` — удалить первое найденное значение;
- `pop()` — удалить и вернуть последний элемент;
- `pop(index)` — удалить элемент по индексу;
- `len(list)` — узнать количество элементов;
- `value in list` — проверить наличие;
- `list[index] = value` — заменить элемент.

Пример:


````python
cart = ["Мышь", "Клавиатура"]

cart.append("Наушники")
cart.insert(1, "Коврик")
print("После добавления:", cart)

cart.remove("Мышь")
print("После remove:", cart)

removed_item = cart.pop()
print("Удалили через pop:", removed_item)
print("Корзина:", cart)

cart[0] = "Игровой коврик"
print("После замены:", cart)

print("Товаров в корзине:", len(cart))
print("Есть клавиатура:", "Клавиатура" in cart)
````

## 3. `set`: уникальные категории

`set` хранит только уникальные значения.

Это удобно, если в каталоге много товаров, но мы хотим узнать,
какие категории вообще представлены.


````python
categories = ["Ноутбуки", "Аксессуары", "Аксессуары", "Мониторы", "Аксессуары"]

unique_categories = set(categories)

print(unique_categories)
````

### Основные операции с `set`

Используем:

- `add(value)` — добавить значение;
- `remove(value)` — удалить существующее значение;
- `discard(value)` — удалить без ошибки, если значения нет;
- `len(set)` — количество уникальных значений;
- `value in set` — проверка наличия;
- `set_a | set_b` — объединение;
- `set_a & set_b` — пересечение.

Пример:


````python
online_categories = {"Ноутбуки", "Мониторы", "Аксессуары"}
store_categories = {"Ноутбуки", "Телевизоры", "Аксессуары"}

print("Все категории:", online_categories | store_categories)
print("Есть в обоих каналах:", online_categories & store_categories)

online_categories.add("Смартфоны")
online_categories.discard("Фотоаппараты")

print("После изменений:", online_categories)
````

## 4. `dict`: карточка товара

`dict` хранит данные в формате `ключ: значение`.

Так удобно описывать один товар.


````python
product = {
    "name": "Ноутбук",
    "price": 79990,
    "category": "Ноутбуки",
    "in_stock": True,
}

print(product)
print("Название:", product["name"])
print("Цена:", product["price"])
````

### Основные операции со словарём

Можно:

- получить значение: `product["price"]`;
- безопасно получить значение: `product.get("discount")`;
- изменить значение: `product["price"] = 74990`;
- добавить ключ: `product["brand"] = "Example"`;
- проверить ключ: `"price" in product`;
- получить ключи: `product.keys()`;
- получить значения: `product.values()`;
- получить пары: `product.items()`;
- удалить ключ: `product.pop("brand")`.

Пример:


````python
product = {
    "name": "Монитор",
    "price": 24990,
    "category": "Мониторы",
    "in_stock": True,
}

product["price"] = 22990
product["brand"] = "ViewPro"

print("Скидка:", product.get("discount"))
print("Есть цена:", "price" in product)
print("Ключи:", list(product.keys()))
print("Значения:", list(product.values()))

for key, value in product.items():
    print(key, "->", value)
````

## 5. `for`: обрабатываем все товары

Цикл `for` выполняет один блок кода для каждого элемента коллекции.


````python
prices = [79990, 1490, 5990, 24990, 7990]

for price in prices:
    print("Цена:", price)
````

### `range()`

`range()` удобно использовать, когда нужна последовательность чисел.


````python
for number in range(1, 6):
    print("Товар №", number)
````

### `enumerate()`

`enumerate()` даёт одновременно номер и значение.


````python
products = ["Ноутбук", "Мышь", "Клавиатура"]

for number, product_name in enumerate(products, start=1):
    print(f"{number}. {product_name}")
````

### `zip()`

`zip()` объединяет несколько списков по позициям.


````python
products = ["Ноутбук", "Мышь", "Клавиатура"]
prices = [79990, 1490, 5990]

for product_name, price in zip(products, prices):
    print(f"{product_name}: {price} руб.")
````

### Практика 5. ⭐ Обязательная на занятии — Печать каталога

**Цель:** различить обычный `for`, `range`, `enumerate`, `zip` и увидеть накопление результата в цикле.

**Исходные данные:**

```python
products = ["Ноутбук", "Мышь", "Клавиатура", "Монитор", "Наушники"]
prices = [79990, 1490, 5990, 24990, 7990]
```

**Что сделать:**

**A. Обычный `for`**
- выведите каждое название товара.

**B. `enumerate()`**
- выведите нумерованный каталог: `1. Ноутбук`, `2. Мышь` и т.д.

**C. `range()`**
- выведите числа от 1 до количества товаров включительно.

**D. `zip()`**
- выведите строки вида `Ноутбук — 79990 руб.`.

**E. Накопление результата в `for`**
- создайте `total_price = 0`;
- внутри цикла прибавьте к нему все цены.

В этой части сумма намеренно считается циклом, чтобы показать накопление значения. В следующем блоке ту же операцию сравним со встроенной `sum()`.

**Ожидаемый результат:** четыре отдельных вида вывода и `total_price == 120450`.

**Критерий готовности:** каждая часть решена указанным инструментом, а вы можете объяснить, зачем нужны `enumerate()` и `zip()`.


````python
products = ["Ноутбук", "Мышь", "Клавиатура", "Монитор", "Наушники"]
prices = [79990, 1490, 5990, 24990, 7990]

# Практика 5.
# A. Обычный for
for product in products:
  print(product)
# B. enumerate()
print("B")
for number, product_name in enumerate(products, start=1):
    print(f"{number}. {product_name}")
# C. range()
print("C")
for number in range(1,len(products)+1):
  print(number)
# D. zip()
print("D")
for product, price in zip(products, prices):
    print(f"{product} — {price} руб.")
# E. total_price
print("E")
total_price = 0
for price in prices:
    total_price += price
print(total_price)
````

- `range()` нужен, когда ...

**Ответ:**

## 7. `while`: собираем заказ до бесплатной доставки

`while` используем, когда заранее неизвестно, сколько повторений потребуется.

Добавляем товары в заказ по очереди, пока сумма не достигнет порога бесплатной доставки
или пока товары не закончатся.


````python
products = ["Мышь", "Клавиатура", "Наушники", "Монитор"]
prices = [1490, 5990, 7990, 24990]
free_delivery_limit = 10000

total = 0
index = 0

while total < free_delivery_limit and index < len(prices):
    total += prices[index]
    print(f"Добавили: {products[index]} — сумма {total} руб.")
    index += 1

if total >= free_delivery_limit:
    print("Бесплатная доставка доступна")
else:
    print("Товары закончились раньше достижения порога")
````
