### Практика 3. ⭐ Разбиваем обработку каталога на функции

**Цель:** декомпозировать знакомый анализ каталога.

**Исходные данные:** используйте `catalog` из примера выше.

**Что сделать:**
1. создайте `get_prices(catalog)`, возвращающую список цен;
2. создайте `average_price(catalog)`, возвращающую среднюю цену;
3. создайте `available_names(catalog)`, возвращающую названия товаров в наличии;
4. вызовите все три функции;
5. выведите результаты.

**Ожидаемый результат:** отдельные список цен, средняя цена и список доступных товаров.

**Критерий готовности:** каждая функция получает `catalog` параметром и возвращает значение; вычисления не зависят от жёстко заданной глобальной переменной внутри функции.

```python
# Практика 3.
def get_prices(catalog):
  return [product["price"] for product in catalog]

def average_price(catalog):
  prices = get_prices(catalog)
  if len(prices) == 0:
    return 0
  else:
    return sum(prices) / len(prices)
def available_names(catalog):
  return [product["name"]for product in catalog if product["in_stock"] == True]

print(get_prices(catalog))
print(average_price(catalog))
print(available_names(catalog))
```

## 6. Обработка ожидаемых ошибок

Иногда данные нельзя преобразовать в число. Вместо падения программы можно обработать ожидаемую `ValueError`.

```python
def parse_price(value):
    try:
        return int(value)
    except ValueError:
        return None

print(parse_price("5990"))
print(parse_price("нет цены"))
```

## 10. Итог: функция загрузки + функция анализа

Теперь отделим получение данных от их анализа.

Одна функция читает CSV и возвращает каталог. Другая принимает любой каталог такой структуры и возвращает сводку.

### Практика 7. ⭐ Главная итоговая практика

**Цель:** собрать небольшой повторно используемый pipeline `файл → данные → сводка → выводы`.

**Исходные данные:** `products.csv`.

**Что сделать:**
1. создайте `load_csv_catalog(filename)`, которая читает CSV и возвращает нормализованный список словарей;
2. создайте `catalog_summary(catalog)`;
3. `catalog_summary` должна вернуть словарь с полями:
   - `count`;
   - `min_price`;
   - `max_price`;
   - `average_price`;
   - `available_count`;
   - `categories`;
4. вызовите загрузчик для `"products.csv"`;
5. передайте результат в `catalog_summary`;
6. выведите сводку;
7. напишите минимум 3 содержательных вывода.

**Ожидаемый результат:** словарь сводки для 6 товаров; вычисления выполняются на данных, возвращённых функцией загрузки.

**Критерий готовности:** функции не используют жёстко заданный каталог внутри себя; загрузчик возвращает данные, анализатор получает их параметром и возвращает результат.

```python
import csv

def load_csv_catalog(filename):
    catalog = []
    with open(filename, encoding="utf-8", newline="") as file:
        reader = csv.DictReader(file)
        for row in reader:
            product = {
                "name": row["name"],
                "price": int(row["price"]),
                "category": row["category"],
                "in_stock": row["in_stock"] == "True"
            }
            catalog.append(product)
    return catalog


def catalog_summary(catalog):
    prices = [product["price"] for product in catalog]
    categories = list(set(product["category"] for product in catalog))

    summary = {
        "count": len(catalog),
        "min_price": min(prices),
        "max_price": max(prices),
        "average_price": sum(prices) / len(prices),
        "available_count": sum(1 for product in catalog if product["in_stock"]),
        "categories": categories
    }
    return summary


catalog = load_csv_catalog("products.csv")
summary = catalog_summary(catalog)


print(summary)
```

## Шаг 3. Сводка по категориям

**Что сделать:** создайте `category_summary(catalog)`, которая возвращает словарь `категория -> количество товаров`.

**Ожидаемый результат:** например, категория «Аксессуары» имеет несколько товаров.

**Критерий готовности:** функция строит результат из переданного `catalog`, не из заранее записанных чисел.
