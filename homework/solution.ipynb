## 7. Самостоятельная практика «Итог заказа»

**Исходные правила:** скидка 10% при сумме от 7000; скидка 5% при сумме
от 5000 и наличии клубной карты; иначе скидка 0%. Доставка бесплатна, если
сумма после скидки не меньше 5000, иначе доставка стоит 300 рублей.

**Что сделать:**

1. Создайте `product_name`, `price`, `quantity`, `is_member`.
2. Рассчитайте `order_total`.
3. Через `if` / `elif` / `else` присвойте `discount_rate`.
4. Рассчитайте `total_after_discount`.
5. Через второе условие присвойте `actual_delivery_cost`.
6. Рассчитайте `final_total` и выведите все результаты одной f-строкой.
7. Проверьте три сценария: без скидки; со скидкой 5%; ровно 7000 до скидки.
8. Для каждого сценария заполните подготовленное поле ниже.

**Готово, если:** код выполняется без ошибок, значения меняются внутри
ветвей, проверены три сценария и вывод соответствует результату.

````python
# 1. Исходные данные
product_name = "Ковёр"
price = 700
quantity = 10
is_member = False
# 2. Исходная сумма
order_total = price * quantity

# 3. Выбор discount_rate
if order_total >= 7000:
    discount_rate = 0.10
elif order_total >= 5000 and is_member:
    discount_rate = 0.05
else:
    discount_rate = 0.0

# 4. Сумма после скидки
total_after_discount = order_total * (1 - discount_rate)

# 5. Выбор actual_delivery_cost
if total_after_discount >= 5000:
    actual_delivery_cost = 0
else:
    actual_delivery_cost = 300

# 6. Итоговая сумма и f-строка
final_total = total_after_discount + actual_delivery_cost

output_string = (
    f"Товар: {product_name} | Сумма: {order_total} руб. | "
    f"Скидка: {discount_rate * 100:.0f}% | После скидки: {total_after_discount} руб. | "
    f"Доставка: {actual_delivery_cost} руб. | Итого к оплате: {final_total} руб."
)
print(output_string)
````

````python
# Расширение 2. Клубная карта
order_total = 4500
is_member = True

# Сравните два правила: достаточная сумма AND карта; достаточная сумма OR карта.
threshold = 5000
rule_and = (order_total >= threshold) and is_member
rule_or = (order_total >= threshold) or is_member

print(f"Результат правила AND: {rule_and}")
print(f"Результат правила OR: {rule_or}")
````

**Вывод расширения 2:** Когда правила дают разные результаты и почему? Правила дают разные результаты, когда выполняется только одно из двух условий. В данном тесте сумма заказа (4500) меньше порога (5000), но покупатель является членом клуба (True). Логическое and требует истинности обеих частей, поэтому дает False. Логическому or достаточно выполнения хотя бы одного условия, поэтому оно возвращает True.

````python
# Расширение 3. Промокод
has_promo_code = True
is_member = True

# Сформулируйте правило: можно ли объединять промокод и клубную скидку?
# Присвойте discount_rate и проверьте минимум три сочетания True / False.
# Сформулируем правило: скидки СКЛАДЫВАЮТСЯ, но максимальная суммарная скидка не может превышать 15%.
# Пусть промокод дает 10%, а клубная карта (is_member) дает 7%.
def calculate_discount(has_promo_code, is_member):
    discount_rate = 0.0
    if has_promo_code:
        discount_rate += 0.10
    if is_member:
        discount_rate += 0.07

    # Ограничение совместного применения
    if discount_rate > 0.15:
        discount_rate = 0.15
    return discount_rate

# Проверка 3 сочетаний True / False
print(f"Промокод: True, Карта: True -> Скидка: {calculate_discount(True, True)*100:.0f}%")
print(f"Промокод: True, Карта: False -> Скидка: {calculate_discount(True, False)*100:.0f}%")
print(f"Промокод: False, Карта: True -> Скидка: {calculate_discount(False, True)*100:.0f}%")
````

**Вывод расширения 3:** Какое правило вы реализовали и к чему оно приводит? Какое правило вы реализовали и к чему оно приводит? Я реализовал правило кумулятивного сложения скидок с ограничением верхнего порога (лимита). Оно приводит к тому, что при раздельном использовании покупатель получает честные 10% или 7%, а при одновременном применении срабатывает лимит, защищающий магазин от чрезмерных убытков, ограничивая общую скидку значением 15% вместо 17%.

````python
# Расширение 4. Собственное правило магазина
# Сначала запишите правило комментарием.
# Затем реализуйте его и проверьте минимум три набора данных.
# Правило: Если количество товаров (quantity) >= 5, то предоставляется
# оптовая скидка 15%. Иначе скидка составляет 0%.
def calculate_wholesale_discount(quantity):
    if quantity >= 5:
        discount_rate = 0.15
    else:
        discount_rate = 0.0
    return discount_rate
# Набор 1: Меньше порогового значения (4 штуки)
qty1 = 4
print(f"Количество: {qty1} шт. -> Скидка: {calculate_wholesale_discount(qty1)*100}%")

# Набор 2: Ровно на границе порога (5 штук)
qty2 = 5
print(f"Количество: {qty2} шт. -> Скидка: {calculate_wholesale_discount(qty2)*100}%")

# Набор 3: Больше порогового значения (10 штук)
qty3 = 10
print(f"Количество: {qty3} шт. -> Скидка: {calculate_wholesale_discount(qty3)*100}%")
````

**Вывод расширения 4:** Почему выбраны именно эти проверки? Что они доказали? Выбранные проверки представляют собой классическое тестирование граничных значений: значение чуть меньше границы (4), точно на границе (5) и заведомо больше нее (10). Они доказали, что логическое условие quantity >= 5 работает корректно — скидка в 15% активируется строго начиная с 5 товаров, а при покупке 4 товаров дополнительная скидка не применяется.
