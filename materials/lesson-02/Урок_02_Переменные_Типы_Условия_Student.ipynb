## 6. `if` / `elif` / `else`

Python проверяет ветви сверху вниз и выполняет первую подходящую. После
условия ставится двоеточие, команды внутри блока имеют одинаковый отступ.

````python
order_total = 6500

if order_total >= 7000:
    discount_rate = 0.10
elif order_total >= 5000:
    discount_rate = 0.05
else:
    discount_rate = 0

total_after_discount = order_total * (1 - discount_rate)
print(f"Скидка: {discount_rate:.0%}; после скидки: {total_after_discount:.2f} руб.")
````

## 8. Кратко: `input()` и `while`

`input()` всегда возвращает строку. Для числа требуется преобразование:

```python
quantity = int(input("Количество: "))
```

Этот пример не запускаем через `Run All`, чтобы Notebook не ожидал ввод.
`while` повторяет блок, пока условие истинно; условие должно изменяться.

````python
attempt = 0
max_attempts = 3

while attempt < max_attempts:
    print(f"Попытка {attempt + 1}")
    attempt += 1
````
