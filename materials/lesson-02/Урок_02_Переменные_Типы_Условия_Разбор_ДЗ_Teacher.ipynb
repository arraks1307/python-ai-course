## ⭐ 7. Самостоятельная практика «Итог заказа»

**Исходные правила:** скидка 10% при сумме от 7000; скидка 5% при сумме
от 5000 и наличии клубной карты; иначе скидка 0%. Доставка бесплатна, если
сумма после скидки не меньше 5000, иначе доставка стоит 300 рублей.

**Что сделать:**

1. Создайте `product_name`, `price`, `quantity`, `is_member`.

4. Рассчитайте `total_after_discount`.
5. Через второе условие присвойте `actual_delivery_cost`.
6. Рассчитайте `final_total` и выведите все результаты одной f-строкой.
7. Проверьте три сценария: без скидки; со скидкой 5%; ровно 7000 до скидки.

ветвей, проверены три сценария и вывод соответствует результату.

````python
# 1. Исходные данные для сценария без скидки
product_name = "Мышь"
price = 1000
quantity = 4
is_member = False

# 2. Исходная сумма
order_total = price * quantity

# 3. Скидка: 10% от 7000; 5% от 5000 при наличии карты
if order_total >= 7000:
    discount_rate = 0.10
elif order_total >= 5000 and is_member:
    discount_rate = 0.05
else:
    discount_rate = 0

# 4. Сумма после скидки
total_after_discount = order_total * (1 - discount_rate)

# 5. Бесплатная доставка от 5000 после скидки
if total_after_discount >= 5000:
    actual_delivery_cost = 0
else:
    actual_delivery_cost = 300

# 6. Итог
final_total = total_after_discount + actual_delivery_cost
print(f"{product_name}: до скидки {order_total} руб.; скидка {discount_rate:.0%}; после скидки {total_after_discount:.0f} руб.; доставка {actual_delivery_cost} руб.; итог {final_total:.0f} руб.")

# Сценарий со скидкой 5%: сумма 6000 руб., карта есть.
price = 1500
quantity = 4
is_member = True
order_total = price * quantity
if order_total >= 7000:
    discount_rate = 0.10
elif order_total >= 5000 and is_member:
    discount_rate = 0.05
else:
    discount_rate = 0
total_after_discount = order_total * (1 - discount_rate)
if total_after_discount >= 5000:
    actual_delivery_cost = 0
else:
    actual_delivery_cost = 300
final_total = total_after_discount + actual_delivery_cost
print(f"Сценарий 2: скидка {discount_rate:.0%}; доставка {actual_delivery_cost} руб.; итог {final_total:.0f} руб.")

# Граница: сумма заказа ровно 7000 руб.
price = 1750
quantity = 4
is_member = False
order_total = price * quantity
if order_total >= 7000:
    discount_rate = 0.10
elif order_total >= 5000 and is_member:
    discount_rate = 0.05
else:
    discount_rate = 0
total_after_discount = order_total * (1 - discount_rate)
if total_after_discount >= 5000:
    actual_delivery_cost = 0
else:
    actual_delivery_cost = 300
final_total = total_after_discount + actual_delivery_cost
print(f"Сценарий 3: сумма {order_total} руб.; скидка {discount_rate:.0%}; итог {final_total:.0f} руб.")
````

## ⭐ Эксперимент 4. Конфликт правил доставки

Используйте `order_total = 6500`, скидку 10% и порог бесплатной доставки
6000 рублей. Сравните два правила:

1. право на бесплатную доставку определяется **до скидки**;
2. право на бесплатную доставку определяется **после скидки**.

Рассчитайте две итоговые суммы.

````python
order_total = 6500
discount_rate = 0.10
free_delivery_threshold = 6000
delivery_cost = 400
total_after_discount = order_total * (1 - discount_rate)

# Правило 1: бесплатную доставку определяем до скидки.
if order_total >= free_delivery_threshold:
    delivery_before_discount = 0
else:
    delivery_before_discount = delivery_cost
final_before_rule = total_after_discount + delivery_before_discount

# Правило 2: бесплатную доставку определяем после скидки.
if total_after_discount >= free_delivery_threshold:
    delivery_after_discount = 0
else:
    delivery_after_discount = delivery_cost
final_after_rule = total_after_discount + delivery_after_discount

print(f"После скидки: {total_after_discount:.0f} руб.")
print(f"Порог до скидки: доставка {delivery_before_discount}; итог {final_before_rule:.0f} руб.")
print(f"Порог после скидки: доставка {delivery_after_discount}; итог {final_after_rule:.0f} руб.")
````

❓ **Вывод 4:** после скидки остаётся 5850 руб. Если порог проверяется до скидки, бесплатная доставка даёт итог 5850 руб.; если после — добавляется 400 руб., итог 6250 руб. Первый вариант выгоднее покупателю, второй — магазину. Понятность зависит от того, какое правило магазин явно сообщает заранее.

❓ **Общий вывод:** проверены пороги скидки и доставки, включая точные границы. Порядок проверки доставки влияет на итог не меньше самой скидки: для заказа 6500 руб. разница между правилами составляет 400 руб. Перед покупкой важно знать, применяется ли порог доставки до скидки или после неё.

````python
order_total = 4500
is_member = True
eligible_with_and = order_total >= 5000 and is_member
eligible_with_or = order_total >= 5000 or is_member
print("AND:", eligible_with_and)
print("OR:", eligible_with_or)
````

❓ **Вывод расширения 2:** при 4500 руб. и наличии карты условие с `and` ложно: не выполнена сумма. Условие с `or` истинно, потому что достаточно одного истинного признака — карты. Эти правила дают разные решения.

````python
# Правило для примера: промокод ИЛИ клубная карта дают 10%; скидки не складываются.
discount_rate = 0.10

has_promo_code = True
is_member = True
if has_promo_code or is_member:
    applied_discount = discount_rate
else:
    applied_discount = 0
print(has_promo_code, is_member, applied_discount)

has_promo_code = True
is_member = False
if has_promo_code or is_member:
    applied_discount = discount_rate
else:
    applied_discount = 0
print(has_promo_code, is_member, applied_discount)

has_promo_code = False
is_member = True
if has_promo_code or is_member:
    applied_discount = discount_rate
else:
    applied_discount = 0
print(has_promo_code, is_member, applied_discount)

has_promo_code = False
is_member = False
if has_promo_code or is_member:
    applied_discount = discount_rate
else:
    applied_discount = 0
print(has_promo_code, is_member, applied_discount)
````

❓ **Вывод расширения 3:** в выбранном правиле скидка 10% применяется, если есть промокод или клубная карта. При обоих признаках она всё равно 10%, то есть скидки не суммируются; без обоих признаков скидка 0%.
