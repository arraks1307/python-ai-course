# Урок 10. Как превращать длинную таблицу в понятный ответ

## Зачем изучать группировки

В таблице может быть сотня или миллион строк. Просмотр отдельных записей не даёт быстрого ответа: **в какой группе чаще достигают результата, сколько людей стоит за процентом и не объясняется ли различие составом групп?** Группировка собирает одинаковые категории и рассчитывает для каждой понятные показатели.

Такой приём нужен в обычной работе:
- в обучении — сравнить долю выполнивших задание в группах разного размера;
- в торговле — посчитать заказы и среднюю сумму покупки по категориям;
- в спорте — сравнить долю посещений по секциям, указав число участников;
- в исследовании — проверить, сохраняется ли наблюдаемое различие внутри сопоставимых подгрупп.

Например, «пять человек выполнили задание» мало что говорит без размера группы. Пять из семи и пять из пятидесяти — разные результаты. Сегодня научимся получать **таблицу, которую можно объяснить словами и использовать для решения**, а не просто запускать groupby.

Сначала разберём прозрачный пример на 12 участниках: каждую строку можно просмотреть и пересчитать вручную. Затем применим тот же способ к Titanic: сравним классы билета, пол внутри класса и возрастные группы. На уроке 9 мы подготовили данные; теперь используем их для ответов на вопросы.

**К концу занятия вы сможете:**
1. Выбрать, что считать одной группой и какой показатель отвечает на вопрос.
2. Рассчитать число людей, число успешных исходов, долю и среднее числового показателя.
3. Объяснить строку сводки, проверить её фильтром и заметить потерю записей.
4. Отличить общее различие от различия внутри подгруппы и назвать ограничения.
5. Сохранить результат и повторить знакомые приёмы на своём CSV в дополнительном разделе.

**Маршрут:** понятная маленькая таблица → ручной расчёт → группировка → проверки → Titanic → собственный вопрос. Первые примеры работают без загрузки внешнего файла.

Для будущей модели такие сводки помогают понять данные и сформулировать предположения о признаках. Они ещё не доказывают причину различий и не проверяют качество прогноза на новых данных. Этому посвятим следующие этапы курса.

## 🟨❗ Теория. Как устроена группировка

Три шага: **разделить строки по ключу → посчитать показатель в каждой группе → собрать сводку**. Ключ — признак, одинаковые значения которого объединяют строки. Исходная таблица остаётся отдельной.

Задайте два вопроса до кода: «что будет одной группой?» и «какой показатель отвечает на мой вопрос?». Число людей, число исходов и доля — разные показатели.

### 🟨❗ Теория. Понятная таблица: 12 участников учебной практики

Представим, что преподаватель провёл одно и то же упражнение в утренней и вечерней группах. Нужно узнать, **как различается доля выполнивших упражнение и среднее время работы**. Данные учебные, не результаты наших студентов.

**Одна строка — один участник и одна попытка.** Всего 12 строк: утром 5 человек, вечером 7. Группы разного размера: сравнивать только число выполнений недостаточно. Участники работали очно или онлайн; возраст известен не у всех.

| Столбец | Что означает | Как читать значение |
|---|---|---|
| participant_id | Номер участника в примере | 1 — номер записи, не возраст и не результат |
| team | Учебная группа | «Утро» или «Вечер» |
| format | Формат участия | «очно» или «онлайн» |
| success | Выполнено ли упражнение | 1 — выполнено, 0 — не выполнено; оба значения известны |
| age | Возраст, лет | Число или пропуск; пропуск не равен нулю |
| practice_minutes | Время работы, минут | Числовой показатель, по которому осмысленно считать среднее |

Английские имена оставляем короткими именами столбцов; смысл каждого записан по-русски. `toy` — имя этой учебной таблицы в коде. `None` создаёт неизвестное значение: pandas покажет его как NaN.

**До запуска:** попробуйте пересчитать строки утренней группы. Их 5; значения success — 1, 0, 1, 1, 0. Значит, упражнение выполнили 3 из 5, то есть 60%. Их время 25, 40, 30, 35, 45 минут: сумма 175, среднее 35 минут.

Следующая ячейка сначала покажет **всю таблицу с 12 строками**, затем по одному рассчитает число выполнений и число участников по группам. `groupby("team")` задаёт группы; `["success"].sum()` складывает единицы внутри каждой; `size()` считает строки. Исходная таблица не исчезает.


```python
import pandas as pd
from pathlib import Path
from IPython.display import display

toy = pd.DataFrame({
    "participant_id": list(range(1, 13)),
    "team": ["Утро"] * 5 + ["Вечер"] * 7,
    "format": ["очно", "очно", "онлайн", "онлайн", "очно",
               "очно", "очно", "очно", "онлайн", "онлайн", "онлайн", "онлайн"],
    "success": [1, 0, 1, 1, 0, 1, 1, 0, 1, 1, 0, 1],
    "age": [20, None, 22, 21, None, 24, 23, None, 25, 26, 24, 22],
    "practice_minutes": [25, 40, 30, 35, 45, 20, 25, 50, 30, 35, 55, 40]
})
display(toy)
successes_by_team = toy.groupby("team")["success"].sum()
display(successes_by_team)
people_by_team = toy.groupby("team").size()
display(people_by_team)
```

**Проверяем:** Утро — 5 участников и 3 выполнения; Вечер — 7 участников и 5 выполнений. Названия групп слева — подписи строк сводки, то есть индекс. В toy остаются все 12 участников.

### 🟨❗ Теория. Несколько показателей одной группировкой

Теперь соберём ответ в одной таблице: сколько участников, сколько выполнили упражнение, какова доля и сколько в среднем работали. Это **разные показатели**; имя столбца должно объяснять, какой из них перед нами.

`agg` собирает несколько расчётов. Слева — выбранное нами имя результата, справа — пара «исходный столбец, действие»:
- `n=("success", "size")`: число участников, то есть строк группы;
- `successes=("success", "sum")`: число единиц — выполнивших;
- `rate=("success", "mean")`: среднее 0/1 — доля выполнивших;
- `mean_minutes=("practice_minutes", "mean")`: среднее время в минутах. Это **не доля**, потому что исходные числа здесь не 0/1.

Имена слева выбираем сами. Имена исходных столбцов справа должны точно совпадать с таблицей. Пару объединяют круглые скобки; кавычки обозначают названия, а не переменные.

Для утренней группы вручную: 3/5=0,6. Умножаем долю на 100 и получаем 60%. Среднее время равно 175/5=35 минут. Следующая ячейка сначала покажет долю 0…1, затем добавит процент 0…100.


```python
summary_example = toy.groupby("team").agg(
    n=("success", "size"),
    successes=("success", "sum"),
    rate=("success", "mean"),
    mean_minutes=("practice_minutes", "mean")
)
display(summary_example)
summary_example["percent"] = summary_example["rate"] * 100
display(summary_example)
```

**Проверяем:** Утро: n=5, successes=3, rate=0,6, percent=60, mean_minutes=35. Вечер: n=7, successes=5, rate≈0,7143, percent≈71,43, mean_minutes≈36,43. Среднее время не является процентом. Порядок строк может быть алфавитным: ищите группу по подписи, а не по номеру строки.

### 🟨❗ Полезный приём: число и доля — разные вопросы

В A выжили 40 из 100 — 40%; в B выжили 30 из 50 — 60%. В A больше выживших людей, в B выше доля. Сначала сформулируйте, что сравниваете. Посчитайте эти два процента вручную; новый код не требуется.

### 🟨❗ Полезный приём: раскройте одну строку сводки

Группировка не должна быть «чёрным ящиком». Проверим строку маленькой таблицы обычным фильтром, затем тем же способом вы проверите первый класс в P1.

```python
morning = toy.loc[toy["team"] == "Утро", :]
display(morning)
manual_n = len(morning)
manual_successes = morning["success"].sum()
manual_rate = manual_successes / manual_n
manual_minutes = morning["practice_minutes"].sum() / manual_n
print("Утро — участников:", manual_n, "Выполнили:", manual_successes)
print("Доля:", manual_rate, "Среднее время:", manual_minutes)
print("Сверьте эти четыре числа со строкой «Утро» в summary_example.")
```

**Проверка после загрузки Titanic:** посмотрите имя открытого файла и сообщение о личном этапе/резерве. Ожидаем 891 пассажира, 342 выживших, 177 неизвестных возрастов. Ниже будет краткий осмотр таблицы: первые строки, размеры и имена столбцов.

```python
# Настройки загрузки Titanic. Меняйте только значения справа.
DATA_FILE = "titanic.csv"       # Точное имя файла или путь, например "data/titanic.csv".
UPLOAD_IN_COLAB = False          # True — выбрать исходник кнопкой в Colab.
PREVIOUS_FILE = None             # None — найти titanic_clean.csv рядом или в data.
SOURCE_SEPARATOR = ","          # Для CSV из Excel иногда нужен ";".
```

```python
from pathlib import Path
import re
import json
import hashlib
import pandas as pd
from IPython.display import display

SOURCE_COLUMNS = ["PassengerId", "Survived", "Pclass", "Name", "Sex", "Age",
                  "SibSp", "Parch", "Ticket", "Fare", "Cabin", "Embarked"]

def read_source(path):
    table = pd.read_csv(path, encoding="utf-8-sig", sep=SOURCE_SEPARATOR)
    missing = set(SOURCE_COLUMNS) - set(table.columns)
    if missing:
        raise ValueError("Не найдены поля: " + ", ".join(sorted(missing)))
    if len(table) != 891 or not table["PassengerId"].is_unique:
        raise ValueError("Нужны 891 запись и уникальные PassengerId.")
    if set(table["PassengerId"]) != set(range(1, 892)):
        raise ValueError("Состав PassengerId отличается от учебного набора.")
    if not table["Survived"].isin([0, 1]).all() or table["Survived"].sum() != 342:
        raise ValueError("Ожидались исходы 0/1 и 342 выживших.")
    if not table["Pclass"].isin([1, 2, 3]).all() or not table["Sex"].isin(["female", "male"]).all():
        raise ValueError("Неожиданные значения Pclass или Sex.")
    if table["Age"].isna().sum() != 177 or not table["Age"].dropna().between(0, 100).all():
        raise ValueError("Возраст отличается от ожидаемого учебного набора.")
    return table[SOURCE_COLUMNS].copy()

# Простой выбор файла: только указанный путь, без поиска папок автора курса.
if UPLOAD_IN_COLAB:
    try:
        from google.colab import files
    except ImportError:
        raise ValueError("Кнопка доступна в Colab. В Jupyter укажите DATA_FILE и поставьте UPLOAD_IN_COLAB = False.")
    uploaded = files.upload()
    if len(uploaded) != 1:
        raise ValueError("Выберите ровно один исходный CSV Titanic. Личный этап можно загрузить отдельно в панель Файлы.")
    DATA_FILE = next(iter(uploaded))
raw_path = Path(DATA_FILE)
if not raw_path.is_file():
    raise ValueError(f"Не найден файл {DATA_FILE}. Рабочая папка: {Path.cwd()}. "
                     "Загрузите CSV, скопируйте его путь в DATA_FILE или включите кнопку Colab.")
try:
    raw = read_source(raw_path)
except (ValueError, OSError, UnicodeError, pd.errors.ParserError) as error:
    raise ValueError("Файл открыт, но не подходит для обязательной практики Titanic: " + str(error) +
                     " Для своей таблицы используйте раздел «Свой CSV». Проверьте также SOURCE_SEPARATOR.")
print("Открыт исходник:", raw_path.name)

def restore_features(table):
    result = table.copy()
    result["Family_size"] = result["SibSp"] + result["Parch"] + 1
    result["Is_alone"] = result["Family_size"] == 1
    result["Age_group"] = "Неизвестный"
    result.loc[result["Age"] < 18, "Age_group"] = "Младше18"
    result.loc[result["Age"] >= 18, "Age_group"] = "От18"
    result["Age_known"] = result["Age"].notna()
    result["Embarked_for_analysis"] = result["Embarked"].fillna("Неизвестно")
    return result

df = restore_features(raw)
previous_paths = ([Path(PREVIOUS_FILE)] if PREVIOUS_FILE else
                  [Path("titanic_clean.csv"), Path("data/titanic_clean.csv")])
previous_path = next((p for p in previous_paths if p.is_file()), None)
previous_used = False
if previous_path:
    try:
        previous = pd.read_csv(previous_path, encoding="utf-8-sig")
        pd.testing.assert_frame_equal(
            previous[SOURCE_COLUMNS].sort_values("PassengerId").reset_index(drop=True),
            raw.sort_values("PassengerId").reset_index(drop=True), check_dtype=False)
        features = ["Family_size", "Is_alone", "Age_group", "Age_known", "Embarked_for_analysis"]
        pd.testing.assert_frame_equal(
            previous.set_index("PassengerId")[features].sort_index(),
            df.set_index("PassengerId")[features].sort_index(), check_dtype=False)
        df = previous.copy()
        previous_used = True
        print("Продолжаем личный результат:", previous_path)
    except (KeyError, ValueError, AssertionError, OSError, UnicodeError, pd.errors.ParserError) as error:
        print("Личный CSV требует исправления; файл не изменён.")
        print("Причина:", str(error).split("\n")[0])
if not previous_used:
    print("Используем учебный резерв из исходника. Это НЕ сдача и НЕ зачёт урока 9.")
    print("Свою работу предыдущего урока восстановите отдельно.")
stage_result = None
print("Пассажиров:", len(df), "Выжили:", int(df["Survived"].sum()),
      "Неизвестный возраст:", int(df["Age"].isna().sum()))
print("Размер таблицы:", df.shape)
print("Столбцы:", list(df.columns))
display(df.head(5))
```

### 🟨❗ Полезный приём: проверьте, не исчезли ли записи

Теперь другая ситуация: неизвестен не возраст, а **ключ группировки**. В отдельной копии toy уберём название команды у последнего участника. Человек есть, но его команда неизвестна.

По умолчанию groupby не создаёт группу для пропущенного ключа. `dropna=False` просит сохранить такую группу. Это не то же самое, что count: сначала решаем, какие строки входят в группы, затем — что считать в каждой группе.

```python
missing_key = toy.copy()
missing_key.loc[11, "team"] = None
usual_sizes = missing_key.groupby("team").size()
all_sizes = missing_key.groupby("team", dropna=False).size()
display(usual_sizes)
display(all_sizes)
print("Исходных строк:", len(missing_key))
print("В группах по умолчанию:", usual_sizes.sum())
print("Включая неизвестную команду:", all_sizes.sum())
```

Ожидаем 12 исходных строк, 11 в группах по умолчанию и 12 при dropna=False. Если сумма размеров не совпала с ожидаемой выборкой, сначала проверьте фильтр и пропуски в ключах.

В Titanic Age_group уже содержит текст «Неизвестный», а не пропущенный ключ. Поэтому эта группа сохраняется. Для расчёта только по известным возрастам сравниваем сумму n с 714, а не с 891: выборку мы сузили намеренно.

## Короткий словарь

`groupby` — образовать группы; `agg` — собрать показатели; `size` — число строк; `count` — число известных значений выбранного поля; `sum` — сумма; `mean` — среднее. `rate` — доля 0…1, `rate_pct` — процент 0…100. `n` — размер группы.

`.loc[условие, :]` выбирает строки и все столбцы; `.notna()` проверяет известность; `.copy()` создаёт отдельную рабочую копию; `.sort_index()` сортирует подписи строк. `.to_csv` сохраняет CSV, `pd.read_csv` читает его. `Path("data").mkdir(exist_ok=True)` создаёт папку, если её ещё нет; `.write_text(..., encoding="utf-8")` записывает русский текст в файл.

`assert` — техническая проверка: если условие неверно, выполнение останавливается. Не удаляйте её ради успешного запуска, а выясните причину. `pd.testing.assert_frame_equal` сравнивает две таблицы. `.round(2)` округляет показ, но не повышает точность данных.
